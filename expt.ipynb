```python
from pathlib import Path
import json

folder_path = path / 'data/interim'
output_path = path / 'data/processed'
merged_data = []
count = 0
# Automatically find all JSON files in the folder
for file_path in folder_path.glob('*.json'):
    try:
        with open(file_path, 'r', encoding='utf-8-sig') as file:
            data = json.load(file)
            if isinstance(data, list):
                count += 1
                merged_data.extend(data)  # Adds items from the list
                print(f"file {count}")
            else:
                merged_data.append(data)   # Safe fallback if it's a single object
    except json.JSONDecodeError:
        print(f"Skipping broken or empty file: {file_path}")

# Save the combined data into a new file
output_path = output_path / 'merged_output.json'
with open(output_path, 'w', encoding='utf-8') as file:
    json.dump(merged_data, file, ensure_ascii=False, indent=4)

print(f"length of all content is {len(merged_data)}")

print(f"Successfully merged files! Total items: {len(merged_data)}")
```

```python
import json
import random

# 1. Load your original JSON file (containing the 2,000 items)
input_path = "D:/marwari-llama3.2-sft/data/processed/marwari_sft.json"  # Update with your actual path
output_path = "D:/marwari-llama3.2-sft/data/processed/subset_150.json"   # Update with your desired output path

with open(input_path, "r", encoding="utf-8") as f:
    data = json.load(f)

# Optional safety check: ensure the data is a list and has enough items
if isinstance(data, list):
    # 2. Randomly select 150 items
    sample_size = min(150, len(data))  # Prevents errors if list has fewer than 150 items
    random_subset = random.sample(data, sample_size)

    # 3. Save the new list to a new JSON file
    with open(output_path, "w", encoding="utf-8") as f:
        # indent=4 makes it pretty-printed and human-readable
        json.dump(random_subset, f, ensure_ascii=False, indent=4)

    print(f"Successfully saved {len(random_subset)} random items to {output_path}")
else:
    print("Error: The root of the JSON file is not a list.")
```
